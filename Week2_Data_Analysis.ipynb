```python
# Check how many corrupted rows exist
print("Before cleaning:", df_combined.shape[0])

# Remove rows where Flow Duration is negative
df_combined = df_combined[df_combined[' Flow Duration'] >= 0]

# Check how many rows remain
print("After cleaning:", df_combined.shape[0])
```

```python
from sklearn.preprocessing import LabelEncoder

# Create encoder
le = LabelEncoder()

# Convert labels to numbers
df_combined[' Label'] = le.fit_transform(df_combined[' Label'])

# Check the result
print(df_combined[' Label'].value_counts())
print("\nLabel mapping:")
for i, label in enumerate(le.classes_):
    print(f"{label} → {i}")
```

```python
# Separate features and labels
X = df_combined.drop(' Label', axis=1)
y = df_combined[' Label']

print("Features shape:", X.shape)
print("Labels shape:", y.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Create scaler
scaler = StandardScaler()

# Scale the features
X_scaled = scaler.fit_transform(X)

print("Scaling done!")
print("First row sample:", X_scaled[0][:5])
```

```python
import numpy as np

# Replace infinity values with NaN first
X = X.replace([np.inf, -np.inf], np.nan)

# Check how many NaN values exist now
print("NaN values found:", X.isnull().sum().sum())

# Drop rows with NaN values
X = X.dropna()
y = y[X.index]

print("After cleaning:")
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Create scaler
scaler = StandardScaler()

# Scale the features
X_scaled = scaler.fit_transform(X)

print("Scaling done!")
print("First row sample:", X_scaled[0][:5])
```

```python
from sklearn.model_selection import train_test_split

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

print("Training set size:", X_train.shape)
print("Testing set size:", X_test.shape)
```

```python
# Compare normal vs attack traffic on Flow Duration
plt.figure(figsize=(10,5))

sns.histplot(data=df_combined[df_combined[' Label']==0][' Flow Duration'], 
             color='green', label='BENIGN', bins=50, alpha=0.6)
sns.histplot(data=df_combined[df_combined[' Label']==1][' Flow Duration'], 
             color='red', label='FTP-Patator', bins=50, alpha=0.6)

plt.title('Flow Duration: Normal vs Attack Traffic')
plt.xlabel('Flow Duration')
plt.ylabel('Count')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Take a sample so both classes are visible
benign_sample = df_combined[df_combined[' Label']==0].sample(7938, random_state=42)
attack_sample = df_combined[df_combined[' Label']==1]

# Combine samples
df_sample = pd.concat([benign_sample, attack_sample])

# Plot
plt.figure(figsize=(10,5))
sns.boxplot(data=df_sample, 
            x=' Label', 
            y=' Flow Duration',
            palette=['green', 'red'])

plt.title('Flow Duration: Normal vs Attack (Balanced Sample)')
plt.xlabel('Label (0=BENIGN, 1=FTP-Patator)')
plt.ylabel('Flow Duration')
plt.tight_layout()
plt.show()
```
